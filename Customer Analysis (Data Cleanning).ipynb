```python
for i in ('Purchase Amount (USD)','Review Rating','Previous Purchases'):
    df[i]=df[i].fillna(df[i].mean())
df[i]
```

```python
col=df.select_dtypes(include='object')
for j in col:
    df[j]=df[j].fillna(df[j].mode()[0])
df[j]
```

```python
cols=['Age','Purchase Amount (USD)','Review Rating','Previous Purchases']
plt.boxplot(df[cols])
plt.xticks(ticks=range(1,len(cols)+1),labels=cols)
```
